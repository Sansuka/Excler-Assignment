##### 1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
lst = [1,2,3,4,5]
print("Original List: ", lst)
lst.append(6)
print("After append(6): ", lst)
lst.extend([7])
print("After extend([7]): ", lst)
lst.insert(2,2.5)
print("After Insert(2,2.5): ", lst)
lst.remove(2.5)
print("After remove(2.5): ", lst)
lst.pop()
print("After pop(): ", lst)
print("Index of 4: ", lst.index(4))
lst.extend([6,8,6,11])
print("count of 6: ", lst.count(6))
lst.sort()
print("After Sort(): ", lst)
lst.reverse()
print("After Reverse(): ", lst)
lst.clear()
print("After clear(): ", lst)
```

##### 7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
n = int(input("Enter a number: "))
if n > 0:
    if n % 2 == 0:
        print("The number is positive and even")
    else:
        print("The number is positive and odd")
elif n == 0:
    print("The number is zero")
else:
    print("The number is negative")
```

#### Section C: Exploring Pandas

##### 11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
data = pd.DataFrame({
    'Name':['Rakshith','Akhil','Amal','Sanskar','Wrishab','Sharan','Rupesh','Siddarth','Tharun','Rakshith'],
    'Age': [23,24,25,21,25,21,21,22,23,23],
    'City':['Bangalore','Hyderabad','Thiruvananthapuram','Puri','Cuttack','Puducherry','Mysuru','Coimbatore','Kochi','Bangalore'],
    'Salary':[90000,80000,85000,90000,80000,np.nan,70000,np.nan,87000,90000]
})
print("Original DataFrame:\n")
print(data)
print("\nDataframe Info:\n")
print(data.info())
print("\nData Description:\n")
print(data.describe())
print("\nName & City Column:\n")
print(data[['Name','City']])
data = data.drop("City", axis=1)
print("\nAfter dropping city column\n")
print(data)
data['Salary'] = data['Salary'].fillna(data['Salary'].mean())
print("\nAfter filling missing Salary with mean\n")
print(data)
duplicates = data[data.duplicated()]
print("\n Duplicates\n",duplicates)
data = data.drop_duplicates()
print("\nAfter removing duplicated rows:")
print(data)
```
