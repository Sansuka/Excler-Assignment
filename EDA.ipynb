### 1.	Data Cleaning and Preparation:
- Load the dataset into a data frame or equivalent data structure.

- Handle missing values appropriately (e.g., imputation, deletion).
- Identify and correct any inconsistencies in data types (e.g., numerical values stored as strings).

```python
cg.duplicated().sum()
```

```python
cg = cg.drop_duplicates()
```

```python
cg.duplicated().sum()
```

```python
cg.isnull().sum()
```

```python
for col in ['AC','FM','UC','DL','DS','DP','ASTV','MSTV','ALTV','MLTV','Width']:
    print(f"\nNegative values in {col}:")
    print((cg[col] < 0).sum())
```

```python
cg.loc[cg['AC'] < 0, 'AC'] = np.nan
cg.loc[cg['FM'] < 0, 'FM'] = np.nan
cg.loc[cg['UC'] < 0, 'UC'] = np.nan
cg.loc[cg['DL'] < 0, 'DL'] = np.nan
cg.loc[cg['DS'] < 0, 'DS'] = np.nan
cg.loc[cg['DP'] < 0, 'DP'] = np.nan
cg.loc[cg['ASTV'] < 0, 'ASTV'] = np.nan
cg.loc[cg['MSTV'] < 0, 'MSTV'] = np.nan
cg.loc[cg['ALTV'] < 0, 'ALTV'] = np.nan
cg.loc[cg['MLTV'] < 0, 'MLTV'] = np.nan
cg.loc[cg['Width'] < 0, 'Width'] = np.nan
```

```python
cg.isnull().sum()
```

```python
mean_cols = ['LB', 'AC', 'UC', 'DS', 'DP', 'ASTV', 'MSTV', 'MLTV', 'Width']
median_cols = ['FM', 'DL', 'ALTV']
mode_cols = ['Tendency']
for col in mean_cols:
    cg[col] = cg[col].fillna(cg[col].mean())
for col in median_cols:
    cg[col] = cg[col].fillna(cg[col].median())
for col in mode_cols:
    cg[col] = cg[col].fillna(cg[col].mode()[0])
```

```python
cg.isnull().sum()
```

```python
cg['NSP'].unique()
```

```python
cg['NSP'] = cg['NSP'].round()
```

```python
cg['NSP'].unique()
```

```python
cg[cg['NSP'].isin([-1, 5])]
```

```python
cg = cg.dropna(subset=['NSP'])
```

```python
cg = cg[cg['NSP'].isin([1,2,3])]
```

```python
cg['NSP'].unique()
```

```python
cg.isnull().sum()
```

```python
cg.info()
```

##### Dataset Overview
- The dataset contains 2126 rows and 14 columns, where each row represents a fetal examination and each column represents a clinical measurement. Using info(), it was observed that all variables are of float64 datatype, meaning all features are numerical and no datatype conversion was required.
- Some columns (LB, AC, DS, DP, MLTV, Width, Tendency, and NSP) contained missing values, which needed to be handled before further analysis.

##### Statistical Summary
The describe() function was used to understand the overall distribution of the data. It provides count, mean, median, mode, standard deviation, minimum, maximum, quartiles (Q1, Q2, Q3), and IQR for each numerical variable.
- Count shows the number of non-missing values.
- Mean represents the average value.
- Median represents the middle value.
- Mode is the most frequently occurring value.
- Standard Deviation measures how much the values vary from the mean.
- Minimum and Maximum show the smallest and largest observations.
- Quartiles divide the data into four equal parts, while IQR (Q3 − Q1) represents the spread of the middle 50% of the data.
##### Observation
For LB, there were 21 missing values. The mean (133.34), median (133), and mode (133) are almost identical, indicating an approximately symmetric distribution. The standard deviation (11.29) suggests moderate variation around the mean. The minimum (51.84) and maximum (214) are extreme observations that require further investigation before deciding whether they are actual outliers. Similar observations were made for the remaining variables.

##### Duplicate Records
- The dataset was checked for duplicate records using duplicated().sum(), and 2 duplicate rows were found. These were removed using drop_duplicates() to avoid duplicate information affecting the analysis.

##### Investigation of Negative Values
- The dataset was checked for negative values in variables where negative measurements are not medically possible. Negative values were found in AC, FM, UC, DL, DS, DP, ASTV, MSTV, ALTV, MLTV, Width, and NSP. Tendency was excluded because negative values are valid for this variable.
- These invalid values were replaced with NaN so they could be treated as missing values and handled appropriately.

##### Handling Missing Values
Missing values were filled based on the distribution of each variable.
- Mean imputation was used for LB, AC, UC, DS, DP, ASTV, MSTV, MLTV, and Width because these variables showed little or no skewness.
- Median imputation was used for FM, DL, and ALTV because they were positively skewed.
- Mode imputation was used for Tendency since it is a categorical/encoded variable.
After imputation, all missing values in the feature columns were successfully handled.

##### Cleaning the Target Variable (NSP)
The NSP column represents the target variable with three valid classes:
- 1 – Normal
- 2 – Suspect
- 3 – Pathological

During preprocessing, decimal values were rounded to the nearest integer. Invalid labels such as -1 and 5 were removed, along with rows containing missing NSP values. Finally, only the valid classes 1, 2, and 3 were retained, making the target variable ready for further analysis and classification.
